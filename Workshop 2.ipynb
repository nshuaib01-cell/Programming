```python
import time 
time.sleep(5) #crawl-delay from website 

def scrape():

    url = 'https://quotes.toscrape.com/author/Jane-Austen/'
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    print(soup)

    title = soup.select_one('h3.author-title').text.strip()
    birth_date = soup.select_one('.author-born-date').text.strip()
    description = soup.select_one('.author-description').text.strip()

    print(f"Author: {title}")
    print(f"Born: {birth_date}")
    print(f"Description: {description}")


if __name__ == '__main__':
    scrape()
```
